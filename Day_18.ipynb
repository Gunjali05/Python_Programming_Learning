```python
#01 WAP to arithmetic operation using function
def add(n1,n2):
    res=n1+n2     #addition
    return res
total=add(5,10)
print(total)
```

```python
#02 WAP to arithmetic operation using function
def sub(n1,n2):
    res=n1-n2     #subtraction
    return res
total=sub(50,10)
print(total)
```

```python
#03 WAP to arithmetic operation using function
def mul(n1,n2):
    res=n1*n2    #multiplication
    return res
total=mul(5,10)
print(total)
```

```python
#04 WAP to arithmetic operation using function
def div(n1,n2):
    res=n1/n2     #division
    return res
total=div(100,10)
print(total)
```

```python
#05 WAP to arithmetic operation using function
def rem(n1,n2):
    res=n1%n2     #remainder
    return res
total=rem(5,10)
print(total)
    
```

```python
#06 WAP to arithmetic operation using function
def quo(n1,n2):
    res=n1//n2   #quotient
    return res
total=quo(5,10)
print(total)
```

```python
#07 WAP to arithmetic operation using function
def power(n1,n2):
    res=n1**n2   #power
    return res
total=power(2,4)
print(total)
```

```python
#08 WAP to calculate square root using function
def squ_root(n1):
    res=1/2*n1
    return res
total=squ_root(8)
print(total)
```

```python
#09 WAP to calculate cube root using function
def cube_root(n1):
    res=1/3*n1
    return res
total=cube_root(27)
print(total)
```

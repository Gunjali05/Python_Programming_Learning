```python
#01 WAP to calculate Time calculator
distance = float(input("Enter distance: "))
speed = float(input("Enter speed: "))
time = distance / speed
print("Time =", time)
```

```python
#02 WAP to convert minutes to hours
minutes = int(input("Enter minutes: "))
hours = minutes // 60
remaining = minutes % 60
print(hours, "hours", remaining, "minutes")
```

```python
#03 WAP to Convert Days to Years, Months and Days
days = int(input("Enter days: "))
years = days // 365
months = (days % 365) // 30
remaining = (days % 365) % 30
print(years, "Years")
print(months, "Months")
print(remaining, "Days")
```

```python
#10 WAP to Find Total Seconds
hours = int(input("Hours: "))
minutes = int(input("Minutes: "))
seconds = int(input("Seconds: "))
total = hours * 3600 + minutes * 60 + seconds
print("Total Seconds =", total)
```

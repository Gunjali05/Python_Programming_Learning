```python
#01 WAP to create a function to check whether a number is Palindrome or Not.
def pali_num(n):
    original = n
    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    if original == reverse:
        print("Palindrome Number")
    else:
        print("Not a Palindrome Number")
pali_num(121)
pali_num(125)
```

```python
#02 WAP to create a function to check whether a given number is an Armstrong number or not.
def arm_num(n):
    temp = n
    sum = 0

    while temp > 0:
        digit = temp % 10
        sum = sum + digit ** 3
        temp = temp // 10

    if sum == n:
        print("Armstrong Number")
    else:
        print("Not Armstrong Number")
arm_num(153)       #1^3=1, 5^3=125, 3^3=27, 1+125+27=153
arm_num(145)
```

```python
#03 WAP to create a function to find the smallest among three numbers.

def smallest(a, b, c):
    if a <= b and a <= c:
        print(f"{a} is the smallest number")
    elif b <= a and b <= c:
        print(f"{b} is the smallest number")
    else:
        print(f"{c} is the smallest number")

smallest(10, 20, 30)
smallest(5, 2, 8)
smallest(15, 25, 10)
```

```python
#04 WAP to create a function to check whether a given year is a Leap Year or not.

def leap(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        print(f"{year} is a Leap Year")
    else:
        print(f"{year} is Not a Leap Year")

leap(2024)
leap(2023)
leap(2000)
leap(1900)
```

```python
#05 WAP to create a function to count the number of vowels in a given string.

def vowels(s):
    count = 0

    for i in s:
        if i.lower() in "aeiou":
            count += 1

    print(f"Number of vowels = {count}")

vowels("Hello World")
vowels("Python")
```

```python
#06 WAP to create a function to check whether a given element is present in a list or not.
def search(list1, ele):
    if ele in list1:
        print(f"{ele} is Present in the List")
    else:
        print(f"{ele} is Not Present in the List")

list1 = [10, 20, 30, 40, 50]

search(list1, 30)
search(list1, 100)
```

```python
#07 WAP to create a function to generate the Fibonacci series up to n terms.
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        c = a + b
        a = b
        b = c

fibonacci(10)
```

```python
#08 WAP to create a function to calculate the GCD (Greatest Common Divisor) of two numbers.
def gcd(a, b):
    while b != 0:
        temp = b
        b = a % b
        a = temp

    print("GCD =", a)

gcd(12, 18)
gcd(20, 30)
```

```python
#09 WAP to create a function to calculate the LCM of two numbers.
def lcm(a, b):
    greater = a

    if b > a:
        greater = b

    while True:
        if greater % a == 0 and greater % b == 0:
            print("LCM =", greater)
            break
        greater += 1

lcm(12, 18)
lcm(10, 15)
```

```python
#10 WAP to create a function to sort a list in ascending order without using sort().
def ascending(list1):
    for i in range(len(list1)):
        for j in range(i + 1, len(list1)):
            if list1[i] > list1[j]:
                temp = list1[i]
                list1[i] = list1[j]
                list1[j] = temp

    print(list1)

list1 = [40, 10, 50, 20, 30]
ascending(list1)
```

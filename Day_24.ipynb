```python
#01 WAP to find the sum of all elements in a list using the reduce() function.
import functools
l1 = [3, 1, 6, 4, 8, 9, 6, 10, 4, 7]
functools.reduce(lambda sum, y: sum + y, l1)
```

```python
#02 WAP to find the product (multiplication) of all elements in a list using reduce().
import functools
l1 = [2, 3, 4, 5]
result = functools.reduce(lambda x, y: x * y, l1)
print(result)
```

```python
#03 WAP to find the maximum element in a list using reduce().
import functools
l1 = [10, 45, 20, 90, 35]
result = functools.reduce(lambda x, y: x if x > y else y, l1)
print(result)
```

```python
#05 WAP to concatenate all strings in a list using reduce().
import functools
l1 = ["Python", " ", "is", " ", "Easy"]
result = functools.reduce(lambda x, y: x + y, l1)
print(result)
```

```python
#06 WAP to count the total sum of even numbers in a list using reduce().
import functools
l1 = [1, 2, 3, 4, 5, 6, 7, 8]
result = functools.reduce(lambda x, y: x + y if y % 2 == 0 else x, l1)
print(result)
```

```python
#07 WAP to calculate the factorial of a number using reduce().
import functools
n = 5
result = functools.reduce(lambda x, y: x * y, range(1, n + 1))
print(result)
```

```python
#08 WAP to find the largest string (by length) in a list using reduce().
import functools
l1 = ["Cat", "Elephant", "Dog", "Tiger"]
result = functools.reduce(lambda x, y: x if len(x) > len(y) else y, l1)
print(result)
```

```python
#09 WAP to find the sum of squares of all elements in a list using reduce().
import functools
l1 = [1, 2, 3, 4]
result = functools.reduce(lambda x, y: x + y**2, l1, 0)
print(result)
```

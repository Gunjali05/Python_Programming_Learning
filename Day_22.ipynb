```python
#01 WAP to create a lambda function to calculate the area of a rectangle.
rectangle = lambda l,b: l*b
rectangle(2,5)
```

```python
#02 WAP to create a lambda function to calculate the area of a triangle.
triangle= lambda h,b: 0.5*h*b
triangle(5,7)
```

```python
#03 WAP to create a lambda function to calculate the area of a circle.
circle = lambda r: 3.14*r**2
circle(27)
```

```python
#04 WAP to create a lambda function to calculate the greater number between two numbers 
gtr=lambda x,y: x if x>y else y
gtr(5,20)
```

```python
#05 WAP to create a lambda function to calculate the smaller number between two numbers 
smlr=lambda x,y: x if x<y else y
smlr(5,20)
```

```python
#06 WAP to create a lambda function to check whether a given number is even or odd.
even_odd=lambda n: "Even" if n%2==0 else "Odd"
even_odd(2)
```

```python
#07 WAP to create a lambda function to convert Celsius to Fahrenheit.
fahrenheit = lambda c: (c * 9/5) + 32
fahrenheit(0)
```

```python
#08 WAP to create a lambda function to check whether a number is positive, negative or zero.
check = lambda n: "Positive" if n > 0 else "Negative" if n < 0 else "Zero"
check(-3)
```

```python
#09 WAP to create a lambda function to calculate the simple interest. (Formula: SI = (P × R × T) / 100)
SI = lambda P,R,T: (P*R*T)/100
SI(10000, 5, 2)
```

```python
#10 WAP to create a lambda function to find the minimum of two numbers.
minimum = lambda a, b: a if a < b else b
print(minimum(10, 20))
```

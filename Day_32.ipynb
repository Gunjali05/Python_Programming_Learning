```python
#01 WAP to create a function to find the shortest word from a list of words.
def shortest_word(words):
    return min(words, key=len)

words = ["Python", "Cat", "Elephant", "Dog"]

print("Shortest Word:", shortest_word(words))
```

```python
#02 WAP to create a function to calculate the total cost of books from a list of book prices.
def total_cost(prices):
    return sum(prices)

prices = [250, 320, 180, 450]

print("Total Cost:", total_cost(prices))
```

```python
#03 WAP to Find Longest Word
from functools import reduce
words = ["Apple", "Watermelon", "Cat", "Banana"]
longest = reduce(lambda x, y: x if len(x) > len(y) else y, words)
print(longest)
```

```python
#04 WAP to calculate Multiply of All Odd Numbers
from functools import reduce
numbers = [1, 2, 3, 4, 5]
odd = list(filter(lambda x: x % 2 != 0, numbers))
result = reduce(lambda x, y: x * y, odd)
print(result)
```

```python
#05 WAP to calculate Partial Function for GST
from functools import partial
def bill(amount, gst):
    return amount + amount * gst / 100
gst18 = partial(bill, gst=18)
print(gst18(500))
```

```python
#06 WAP to calculate Sum of Squares
from functools import reduce
numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y**2, numbers, 0)
print(result)
```

```python
#07 WAP to Count Vowels
from functools import reduce
text = "DataAnalyst"
count = reduce(lambda x, y: x + 1 if y.lower() in "aeiou" else x, text, 0)
print(count)
```

```python
#08 WAP to Count Consonants
from functools import reduce
text = "Python"
count = reduce(
    lambda x, y: x + 1 if y.isalpha() and y.lower() not in "aeiou" else x,
    text,
    0
)
print(count)
```

```python
#09 WAP to Find Total Length of Words
from functools import reduce
words = ["Python", "SQL", "Excel"]
length = reduce(lambda x, y: x + len(y), words, 0)
print(length)
```

```python
#10 WAP to Count Positive Numbers
from functools import reduce

numbers = [-2, 5, -7, 8, 10]

count = reduce(lambda x, y: x + 1 if y > 0 else x, numbers, 0)

print(count)
```

```python
#01 WAP to calculate area of square using function
def area_of_squ(s):
    squ=s**2
    return squ
area_of_squ(4)
```

```python
#02 WAP to calculate area of circle using function
def area_of_cir(r):
    cir=3.14*r**2
    return cir
area_of_cir(5)
```

```python
#03 WAP to calculate area of triangle using function
def area_of_tri(h,b):
    tri=0.5*h*b
    return tri
area_of_tri(5,3)
```

```python
#04 WAP to create a function to check whether a number is Even or Odd.
def even_odd(n):
    if(n%2==0):
        print(f"{n} is Even number")
    else:
        print(f"{n} is Odd number")
even_odd(5)
even_odd(2)
```

```python
#05 WAP to create a function to find the largest among three numbers.
def larger_num(a,b,c):
    if a>b:
        print(f"{a} is largest number")
    elif b>c:
        print(f"{b} is largest number")
    else:
        print(f"{c} is largest number")
larger_num(5,8,1)
larger_num(9,1,4)
larger_num(1,2,5)
```

```python
#06 WAP to create a function to check whether a number is Positive or Negative.
def pos_neg(n):
    if(n>0):
        print(f"{n} is positive")
    else:
        print(f"{n} is negative")
pos_neg(5)
pos_neg(-2)
```

```python
#07 WAP to create a function to calculate the factorial of a number.
def fac(num):
    n=num
    res=1
    while(num>0):
        res=res*num
        num=num-1
    print(f"{res} is a factorial of {n}")
fac(5)
fac(2)
```

```python
#08 WAP to create a function to check whether a number is Prime or Not.
def prime_num(num):
    for i in range(2,num):
        if(num%i==0):
            print(f"{num} is not prime number")
            break
    else:
        print(f"{num} is a prime number")
prime_num(5)
prime_num(4)
```

```python
#09 WAP to create a function to calculate the sum of digits of a given number.
def digit(num):
    sum = 0
    while num > 0:
        sum += num % 10
        num //= 10
    print(sum)

digit(45)
```

```python
#10 WAP to create a function to reverse a given number.
def rev_num(n):
    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    print("Reverse Number =", reverse)
rev_num(45)
```

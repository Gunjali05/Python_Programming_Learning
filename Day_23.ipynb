```python
#01 WAP to create a lambda function to convert Fahrenheit to Celsius. (Formula: (F − 32) × 5/9)
celsium = lambda f: (f * 9/5) + 32
celsium(0)
```

```python
#02 WAP to create a lambda function to calculate the circumference of a circle. (Formula: 2 × π × r)
circle = lambda r : (2*3.14*r)
circle(4)
```

```python
#03 WAP to create a lambda function to check whether a year is a leap year or not.
year = lambda y: "Leap Year" if (y % 400 == 0) or (y % 4 == 0 and y % 100 != 0) else "Not a Leap Year"
print(year(2024))
print(year(2025))
```

```python
#04 WAP to create a lambda function to check whether a character is a vowel or a consonant.
char = lambda ch: "Vowel" if ch.lower() in "aeiou" else "Consonant"
print(char("a"))
print(char("b"))
```

```python
#05 WAP to use filter() with a lambda function to extract all even numbers from a list.
l1=[3,1,6,4,8,9,6,10,4,7]
list(filter(lambda x: x%2==0,l1))
```

```python
#06 WAP to use filter() with a lambda function to extract all numbers less than 5 from a list.
list(filter(lambda x: x<5  ,l1))
```

```python
#07 WAP to use filter() with a lambda function to extract all numbers greater than 5 from a list.
list(filter(lambda x: x>5  ,l1))
```

```python
#09 WAP to use map() with a lambda function to calculate the cube of each element in a list.
list(map(lambda x : x**3 , l1))
```

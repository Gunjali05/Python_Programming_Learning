```python
#01 WAP to update dictionary
d3={1: 'Ram', 2: 'sham', 3: 'jay', 4: 'om'}
d3.update({5:'sachin'})
print(d3)
```

```python
#02 WAP to print total items
d3.items()
```

```python
#03 WAP to print keys
d3.keys()
```

```python
#04 WAP to print values
d3.values()
```

```python
#05 WAP to pop item 
d3.popitem()
print(d3)
```

```python
#06 WAP to pop 3rd item
d3.pop(3)
print(d3)
```

```python
#07 WAP to delete a 2nd element
del d3[2]
print(d3)
```

```python
#08 WAP to print clear all elements
d3.clear()
print(d3)
```

```python
#09 WAP to print all keys and values of a dictionary.

student = {
    "Name": "Amit",
    "Age": 20,
    "City": "Pune"
}

for key in student:
    print(key, ":", student[key])
```

```python
#10 WAP to find the sum of all values in a dictionary.

marks = {
    "Math": 85,
    "Science": 90,
    "English": 78,
    "Computer": 95
}

total = 0

for key in marks:
    total = total + marks[key]

print("Total Marks =", total)
```

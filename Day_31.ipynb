```python
#01 WAP to create a function to calculate the average marks of a student from five subjects and display
#whether the student is Pass or Fail (Pass if average is greater than or equal to 35).
def student_result(marks):
    average = sum(marks) / len(marks)

    print("Average Marks:", average)

    if average >= 35:
        print("Result: Pass")
    else:
        print("Result: Fail")

marks = [45, 60, 38, 50, 42]
student_result(marks)
```

```python
#02 WAP to create a function to calculate the total shopping bill from a list of item prices.
def shopping_bill(prices):
    total = sum(prices)
    return total
prices = [250, 120, 80, 450]
bill = shopping_bill(prices)
print("Total Shopping Bill: ₹", bill)
```

```python
#03 WAP to create a function to find the highest temperature from a list of temperatures recorded over seven days.
def highest_temperature(temps):
    highest = max(temps)
    return highest
temps = [30, 32, 29, 35, 31, 33, 34]
result = highest_temperature(temps)
print("Highest Temperature:", result, "°C")
```

```python
#04 WAP to create a function to calculate the attendance percentage of a student using the total classes and attended classes.  
def attendance_percentage(total_classes, attended_classes):
    percentage = (attended_classes / total_classes) * 100
    print("Attendance Percentage:", percentage, "%")
    if percentage >= 75:
        print("Eligible for Exam")
    else:
        print("Not Eligible for Exam")
attendance_percentage(120, 108)
```

```python
#05 WAP to create a function to calculate the final salary of an employee after adding the bonus percentage.
def final_salary(salary, bonus):
    total_salary = salary + (salary * bonus / 100)
    return total_salary
salary = float(input("Enter Salary: "))
bonus = float(input("Enter Bonus Percentage: "))
result = final_salary(salary, bonus)
print("Final Salary: ₹", result)
```

```python
#06 WAP to create a function to count the total number of even numbers in a list.
def count_even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count = count + 1
    return count
numbers = [10, 15, 22, 35, 40, 51, 68, 73]
result = count_even(numbers)
print("Total Even Numbers:", result)
```

```python
#07 WAP to create a function to calculate the final price of a product after applying a discount percentage.
def final_price(price, discount):
    amount = price * discount / 100
    final = price - amount
    return final
price = float(input("Enter Product Price: "))
discount = float(input("Enter Discount Percentage: "))
result = final_price(price, discount)
print("Final Price: ₹", result)
```

```python
#08 WAP to create a function to find the oldest person from a list of ages.
def oldest_person(ages):
    oldest = max(ages)
    return oldest
ages = [18, 42, 35, 60, 27]
result = oldest_person(ages)
print("Oldest Age:", result)
```

```python
#09 WAP to create a function to calculate the total electricity bill.
def electricity_bill(units, rate):
    bill = units * rate
    return bill
units = int(input("Enter Units Consumed: "))
rate = float(input("Enter Rate per Unit: "))
print("Total Electricity Bill: ₹", electricity_bill(units, rate))
```

```python
#10 WAP to create a function to calculate monthly savings.
def monthly_savings(income, expenses):
    savings = income - expenses
    return savings
income = float(input("Enter Monthly Income: "))
expenses = float(input("Enter Monthly Expenses: "))
print("Monthly Savings: ₹", monthly_savings(income, expenses))
```

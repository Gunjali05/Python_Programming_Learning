```python
#01 WAP to Create a function to calculate electricity bill.
def electricity_bill(units):
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = units * 7
    else:
        bill = units * 10
        
    return bill

print("Bill =", electricity_bill(250))
```

```python
#02 WAP to calculate a percentage of 5 subjects
def percentage(m1, m2, m3, m4, m5):
    total = m1+m2+m3+m4+m5
    per = (total/500)*100
    return per

print("Percentage =", percentage(80,75,90,85,70))
```

```python
#03 WAP to create a function to calculate the total salary of an employee using the basic salary, where HRA = 20% and DA = 10% of the basic salary.
def salary(basic):
    hra = basic * 0.20
    da = basic * 0.10
    
    total = basic + hra + da
    
    return total

print("Salary =", salary(30000))
```

```python
#04 WAP to create a function to calculate the discount based on the shopping amount.
def discount(amount):
    if amount >= 5000:
        return amount * 0.20
    elif amount >= 2000:
        return amount * 0.10
    else:
        return 0

print("Discount =", discount(6000))
```

```python
#05 WAP to create a function to calculate the Simple Interest. 
def interest(p, r, t):
    return (p*r*t)/100

print(interest(10000,5,2))
```

```python
#06 WAP to create a function to calculate a person's age from the birth year.
def age(birth_year):
    current_year = 2026
    return current_year - birth_year

print("Age =", age(2004))
```

```python
#07 WAP to create a function to calculate the taxi fare based on the distance traveled.
def taxi_fare(distance):
    fare = distance * 15
    return fare

print("Fare =", taxi_fare(10))
```

```python
#08 WAP to create a function to calculate the employee bonus based on the salary.
def bonus(salary):
    if salary > 50000:
        return salary*0.10
    else:
        return salary*0.05

print(bonus(60000))
```

```python
#09 WAP to create a function to calculate the Body Mass Index (BMI).
def bmi(weight, height):
    result = weight/(height**2)
    
    return result

print(bmi(70,1.75))
```

```python
#10 WAP to create a function to perform an ATM withdrawal and check whether the account has sufficient balance.
def withdraw(balance, amount):
    if amount <= balance:
        return balance - amount
    else:
        return "Insufficient Balance"

print(withdraw(10000,3000))
```
